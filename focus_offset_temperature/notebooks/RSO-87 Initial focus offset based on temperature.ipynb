## Initial Alignment

Let's start by mining the initial alignment.

```python
day_obs_begin = 20260426
day_obs_end = 20260427
```

```python
from astropy.time import Time

from lsst.summit.utils.dateTime import getDayObsStartTime, getDayObsEndTime
from lsst.summit.utils.efdUtils import getEfdData, makeEfdClient
```

```python
# Make an EFD client to fetch data from the EFD.
efd_client = makeEfdClient("usdf_efd")

# efd_client = makeEfdClient("summit_efd")
```

```python
t_begin = getDayObsStartTime(day_obs_begin)
t_end = getDayObsEndTime(day_obs_end)

df_add_block = await efd_client.select_time_series(
    "lsst.sal.ESS.logevent_ringssMeasurement",
    ["fwhmFree"],
    t_begin,
    t_end,
)

# df_add_block = df_add_block[df_add_block["id"] == "BLOCK-T539"]
df_add_block.head()
```
